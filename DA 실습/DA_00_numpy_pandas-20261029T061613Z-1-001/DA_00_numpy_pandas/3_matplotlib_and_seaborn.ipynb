````
# pandas를 pd라는 이름으로 불러오기
import pandas as pd
````

* .read_csv( )

````
# CSV 파일을 읽어 DataFrame으로 저장
url = 'https://raw.githubusercontent.com/rusita-ai/pyData/master/PII.csv'

DF = pd.read_csv(url)

DF.head()
````

# matplotlib Package

````
# 그래프 출력을 위한 matplotlib 불러오기
import matplotlib.pyplot as plt
````

# I. 선 그래프

* https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.plot.html#matplotlib.axes.Axes.plot
 - [Named Colors](https://matplotlib.org/stable/gallery/color/named_colors.html)
 - [RGB](https://htmlcolorcodes.com)
 - [Line Styles](https://matplotlib.org/stable/gallery/lines_bars_and_markers/linestyles.html)

````
# 그래프 크기 설정 후 시각화 시작
# 그래프 크기를 설정하고 새로운 그래프 영역 생성
plt.figure(figsize = (10, 7))

# Height(키) 데이터를 선 그래프로 시각화
plt.plot(DF.Height,           # 그래프로 표현할 Height(키) 데이터
         linewidth = 1,       # 선의 두께
         color = 'r',         # 선 색상 → 빨간색(red)
         marker = '>',        # 각 데이터 위치를 > 모양으로 표시
         linestyle = '--')    # 선 모양 → 점선


plt.title('Line Graph', size = 30)       # 그래프 제목과 글자 크기
plt.xlabel('Index', size = 20)           # X축 이름과 글자 크기
plt.ylabel('Height', size = 20)          # Y축 이름과 글자 크기
plt.grid(True)                           # 격자 표시
plt.show()                               # 완성된 그래프 화면에 출력
````

# II. 막대 그래프

> ## 1) .bar( )

* https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.bar.html#matplotlib.axes.Axes.bar

````
# 그래프 크기 설정 후 시각화 시작

# 그래프 크기를 설정하고 새로운 그래프 영역 생성
plt.figure(figsize = (10, 7))

# Height(키) 데이터를 세로 막대 그래프로 시각화
plt.bar(DF.index,             # X축 → 데이터의 인덱스(행 번호)
        DF.Height,            # Y축 → Height(키) 값
        width = 0.4,          # 막대의 너비
        color = 'y',          # 막대 내부 색상 → 노란색(yellow)
        edgecolor = 'b')      # 막대 테두리 색상 → 파란색(blue)
plt.show()                    # 완성된 그래프 화면에 출력
````

* 막대 그래프 2개

````
# 그래프 크기 설정 후 시각화 시작

# 그래프 크기를 설정하고 새로운 그래프 영역 생성
plt.figure(figsize = (10, 7))

# Height(키) 데이터를 막대 그래프로 표시
plt.bar(DF.index, DF['Height'],           # X축 → 데이터의 인덱스,  Y축 → Height(키)
        width = 0.3, label = 'Height')    # 막대 너비, 범례에 표시할 이름


# Weight(몸무게) 데이터를 Height 막대의 오른쪽에 표시
plt.bar(DF.index + 0.3, DF['Weight'],    # X축 위치를 0.3만큼 오른쪽으로 이동,  Y축 → Weight(몸무게)
        width = 0.3, label = 'Weight')   # 막대 너비, # 범례에 표시할 이름

plt.legend()     # Height, Weight 범례 표시  (범례(Legend) = 그래프의 선이나 막대가 어떤 데이터를 의미하는지 알려주는 이름표)
plt.show()        # 완성된 그래프 화면에 출력

# DF.index + 0.3
# 두 막대가 겹치지 않고 나란히 보이도록 Weight 막대를 오른쪽으로 0.3 이동시킴
````

> ## 2) .barh( )

* https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.barh.html#matplotlib.axes.Axes.barh

````
# 그래프 크기 설정 후 시각화 시작

# 그래프 크기를 설정하고 새로운 그래프 영역 생성
plt.figure(figsize = (7, 10))

# Height(키) 데이터를 가로 막대 그래프로 표시
plt.barh(DF.index,                              # Y축 → 데이터의 인덱스
         width = DF['Height'], height = 0.3,    # 막대의 길이 → Height(키) 값, 막대의 두께
         label = 'Height')                      # 범례에 표시할 이름

# Weight(몸무게) 막대를 Height 막대의 위쪽에 표시
plt.barh(DF.index + 0.3,                        # Y축 위치를 0.3만큼 이동
         width = DF['Weight'], height = 0.3,    # 막대의 길이 → Weight(몸무게) 값, 막대의 두께
         label = 'Weight')                      # 범례에 표시할 이름

plt.legend()     # 범례(데이터 이름표) 표시
plt.show()       # 완성된 그래프 화면에 출력

# barh가 가로(horizontal) 막대 그래프라서 width가 막대의 길이, height가 막대의 두께가 됨
````

# III. 히스토그램

* https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.hist.html#matplotlib.axes.Axes.hist

````
# 그래프 크기 설정 후 시각화 시작

# 그래프 크기를 설정하고 새로운 그래프 영역 생성
plt.figure(figsize = (10, 7))

plt.hist(DF.Height,          # 히스토그램으로 표현할 Height(키) 데이터
         bins = 5,           # 데이터를 5개의 구간으로 나눔
         alpha = 0.5, density = False)    # 막대의 투명도 (0=투명, 1=불투명), 각 구간의 실제 데이터 개수(빈도)를 표시
plt.show()                   # 완성된 그래프 화면에 출력


# density=False는 비율이 아니라 각 구간에 몇 명이 있는지 개수로 보여준다
````

# IV. 상자 그래프

* https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.boxplot.html#matplotlib.axes.Axes.boxplot

> ## 1) 전체 키 분포

````
# 그래프 크기 설정 후 시각화 시작

# 그래프 크기를 설정하고 새로운 그래프 영역 생성
plt.figure(figsize = (10, 7))

# Height(키) 데이터의 분포를 박스플롯(Box Plot)으로 시각화
plt.boxplot(DF.Height,                # 박스플롯으로 표현할 Height(키) 데이터
            patch_artist = True)      # 박스 내부에 색상을 채울 수 있도록 설정
plt.show()                            # 완성된 그래프 화면에 출력

# patch_artist=True → 박스 안쪽에 색상을 넣을 수 있게 해주는 설정
````

> ## 2) 혈액형별 키분포

````
# 행/열 이름(label)을 기준으로 데이터 선택


# 그래프 크기를 설정하고 새로운 그래프 영역 생성
plt.figure(figsize = (10, 7))

# 혈액형별 Height(키) 분포를 박스플롯으로 비교
plt.boxplot([DF.loc[DF['BloodType'] == 'A', 'Height'],       # A형 사람들의 키 데이터 선택
             DF.loc[DF['BloodType'] == 'B', 'Height'],       # B형 사람들의 키 데이터 선택
             DF.loc[DF['BloodType'] == 'O', 'Height'],       # O형 사람들의 키 데이터 선택
             DF.loc[DF['BloodType'] == 'AB', 'Height']],     # AB형 사람들의 키 데이터 선택
             labels = ['A', 'B', 'O', 'AB'],                 # 각 박스의 이름(혈액형)
            patch_artist = True,                             # 박스 내부에 색상을 채울 수 있도록 설정
            showmeans = True)                                # 평균값도 함께 표시

plt.show()      # 완성된 그래프 화면에 출력

# loc
# DF.loc[DF['BloodType'] == 'A', 'Height']
# → 혈액형이 A형인 행만 찾고 → 그 사람들의 Height(키)만 가져온다.
# 혈액형별로 키의 분포를 비교하는 그래프
````

# V. 산점도

* https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.scatter.html#matplotlib.axes.Axes.scatter

````
# 그래프 크기 설정 후 시각화 시작

# 그래프 크기를 설정하고 새로운 그래프 영역 생성
plt.figure(figsize = (10, 7))

# Height(키)와 Weight(몸무게)의 관계를 산점도로 시각화
plt.scatter(DF.Height,        # X축 → Height(키)
            DF.Weight,        # Y축 → Weight(몸무게)
            marker='*',       # 데이터 점을 별(*) 모양으로 표시
            s = 100,          # 점의 크기
            c = '#d62728')    # 점의 색상
plt.show()                    # 완성된 그래프 화면에 출력


# scatter = 산점도, 즉 키와 몸무게처럼 두 데이터가 서로 어떤 관계가 있는지 점으로 확인하는 그래프
````

# VI. 파이 그래프

* https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.pie.html#matplotlib.axes.Axes.pie

````
# 각 값이 몇 번 나타나는지 개수 확인

# 그래프 크기를 설정하고 새로운 그래프 영역 생성
plt.figure(figsize = (10, 10))

# 혈액형별 인원수를 계산하여 원형 그래프로 시각화
plt.pie(DF.BloodType.value_counts(),                       # 혈액형별 인원수 계산
        labels = DF.BloodType.value_counts().index,        # 혈액형 이름(A, B, O, AB) 표시
        autopct = '%.1f%%',                                # 비율을 소수점 첫째 자리까지 %로 표시
        explode = [0.0, 0.0, 0.0, 0.05],                   # 네 번째 조각을 원에서 조금 떨어뜨려 강조
        startangle = 90)                                   # 원형 그래프를 90도 위치에서 시작
plt.show()      # 완성된 그래프 화면에 출력

# explode(익스플로드-폭발하다, 터뜨리다.)는 특정 조각을 원에서 살짝 떼어내 강조하는 옵션
# 0.05가 지정된 네 번째 조각이 조금 튀어나와 보임
````

# VII. plt.subplot( )
- subplot(서브플롯)
- 하나의 Figure(큰 도화지) 안에 들어가는 각각의 작은 그래프 영역
* https://matplotlib.org/stable/gallery/color/named_colors.html
```
Figure = 전체 도화지
Subplot = 도화지 안에 나눠 놓은 각각의 그래프 칸
ex)
plt.subplot(1, 2, 1)
1행 2열로 나누고, 그중 1번째 그래프 영역을 선택

> ## 1) Figure
- Figure(피규어) = 그래프 전체를 담는 큰 도화지

````
# 그래프 출력을 위한 matplotlib 불러오기
import matplotlib.pyplot as plt

# 그래프를 그릴 전체 영역(Figure) 생성
plt.figure(figsize = (3, 3), facecolor = 'tan')    # 그래프 크기 (가로 3, 세로 3),  Figure의 배경색을 tan(황갈색)으로 설정
plt.title('Figure')         # 그래프 제목 설정
plt.xticks([])              # X축의 눈금과 숫자를 숨김
plt.yticks([])              # Y축의 눈금과 숫자를 숨김
plt.plot()                  # 그래프 영역 생성

plt.show()                  # 완성된 그래프 화면에 출력
````

> ## 2) Axes

````
# 그래프를 그릴 영역(Axes)을 생성

# 그래프를 그릴 영역(Axes)을 생성
ax = plt.axes(facecolor = 'g')     # 그래프 영역의 배경색을 초록색(green)으로 설정


# 전체 Figure(도화지)의 크기 설정
ax.figure.set_size_inches(3, 3)

ax.set_title('Axes')     # 그래프 영역의 제목 설정
ax.set_xticks([])        # X축의 눈금과 숫자를 숨김
ax.set_yticks([])        # Y축의 눈금과 숫자를 숨김

plt.show()               # 완성된 그래프 화면에 출력

# Axes(액시즈)
# Figure = 그래프 전체를 담는 큰 도화지
# Axes = 실제 그래프가 그려지는 영역(그래프 칸)
````

> ## 3) <font color='blue'>1 x 2</font> Subplot

````
# 1행 2열의 그래프 영역을 생성
fig, ax = plt.subplots(
    nrows=1,              # 행의 개수 → 1행
    ncols=2,              # 열의 개수 → 2열
    figsize=(6, 3),       # 전체 Figure 크기 (가로 6, 세로 3)
    facecolor='tan'       # 전체 Figure의 배경색
)

# 첫 번째 그래프 영역(Axes) 설정
ax[0].plot()
ax[0].set_facecolor('tomato')  # 첫 번째 그래프 영역의 배경색
ax[0].set_title('Axes_1')      # 첫 번째 그래프 제목
ax[0].set_xticks([])           # X축 눈금과 숫자 숨김
ax[0].set_yticks([])           # Y축 눈금과 숫자 숨김

# 두 번째 그래프 영역(Axes) 설정
ax[1].plot()
ax[1].set_facecolor('royalblue')  # 두 번째 그래프 영역의 배경색
ax[1].set_title('Axes_2')         # 두 번째 그래프 제목
ax[1].set_xticks([])              # X축 눈금과 숫자 숨김
ax[1].set_yticks([])              # Y축 눈금과 숫자 숨김

plt.show()                         # 완성된 그래프 화면에 출력

# fig = 전체 도화지(Figure)
# ax[0] = 첫 번째 그래프 칸(Axes)
# ax[1] = 두 번째 그래프 칸(Axes)
````

> ## 4) <font color='blue'>2 x 1</font> Subplot

````
# 2행 1열의 그래프 영역을 생성
fig, ax = plt.subplots(
    nrows=2,              # 행의 개수 → 2행
    ncols=1,              # 열의 개수 → 1열
    figsize=(3, 6),       # 전체 Figure 크기 (가로 3, 세로 6)
    facecolor='tan'       # 전체 Figure의 배경색
)

# 첫 번째 그래프 영역(Axes) 설정
ax[0].plot()
ax[0].set_facecolor('limegreen')  # 첫 번째 그래프 영역의 배경색
ax[0].set_title('Axes_1')         # 첫 번째 그래프 제목
ax[0].set_xticks([])              # X축 눈금과 숫자 숨김
ax[0].set_yticks([])              # Y축 눈금과 숫자 숨김

# 두 번째 그래프 영역(Axes) 설정
ax[1].plot()
ax[1].set_facecolor('plum')       # 두 번째 그래프 영역의 배경색
ax[1].set_title('Axes_2')         # 두 번째 그래프 제목
ax[1].set_xticks([])              # X축 눈금과 숫자 숨김
ax[1].set_yticks([])              # Y축 눈금과 숫자 숨김

plt.show()                         # 완성된 그래프 화면에 출력


# 2행 × 1열이므로 그래프가 위아래로 2개 배치돼.
# nrows=2 → 세로로 2칸
# ncols=1 → 가로로 1칸
````

> ## 5) <font color='red'>2 x 2</font> Subplot

````
# 2행 2열의 그래프 영역을 생성
fig, ax = plt.subplots(
    nrows=2,              # 행의 개수 → 2행
    ncols=2,              # 열의 개수 → 2열
    figsize=(6, 6),       # 전체 Figure 크기 (가로 6, 세로 6)
    facecolor='tan'       # 전체 Figure의 배경색
)

# 첫 번째 그래프 영역 → 1행 1열
ax[0, 0].plot()
ax[0, 0].set_facecolor('tomato')  # 배경색 설정
ax[0, 0].set_title('Axes_1')      # 그래프 제목
ax[0, 0].set_xticks([])           # X축 눈금과 숫자 숨김
ax[0, 0].set_yticks([])           # Y축 눈금과 숫자 숨김

# 두 번째 그래프 영역 → 1행 2열
ax[0, 1].plot()
ax[0, 1].set_facecolor('royalblue')
ax[0, 1].set_title('Axes_2')
ax[0, 1].set_xticks([])
ax[0, 1].set_yticks([])

# 세 번째 그래프 영역 → 2행 1열
ax[1, 0].plot()
ax[1, 0].set_facecolor('limegreen')
ax[1, 0].set_title('Axes_3')
ax[1, 0].set_xticks([])
ax[1, 0].set_yticks([])

# 네 번째 그래프 영역 → 2행 2열
ax[1, 1].plot()
ax[1, 1].set_facecolor('plum')
ax[1, 1].set_title('Axes_4')
ax[1, 1].set_xticks([])
ax[1, 1].set_yticks([])

plt.show()  # 완성된 그래프 화면에 출력
````

# VIII. Examples

> ## 1) Histograms

````
# 1행 2열의 그래프 영역을 생성
fig, ax = plt.subplots(
    nrows=1,             # 행의 개수 → 1행
    ncols=2,             # 열의 개수 → 2열
    figsize=(15, 5)      # 전체 Figure 크기 (가로 15, 세로 5)
)

# 첫 번째 그래프 → 실제 데이터 개수(빈도)를 히스토그램으로 표시
ax[0].hist(
    DF.Height,           # Height(키) 데이터
    bins=5,              # 데이터를 5개의 구간으로 나눔
    alpha=0.3,           # 막대의 투명도
    density=False        # 각 구간의 실제 데이터 개수(빈도)를 표시
)

# 두 번째 그래프 → 확률밀도 형태로 히스토그램 표시
ax[1].hist(
    DF.Height,           # Height(키) 데이터
    bins=5,              # 데이터를 5개의 구간으로 나눔
    alpha=0.7,           # 막대의 투명도
    density=True         # 데이터 개수가 아닌 확률밀도(Density)로 표시
)

# 각 그래프의 제목 설정
ax[0].set_title('Histogram-1', size=20)
ax[1].set_title('Histogram-2', size=20)

# X축 이름 설정
ax[0].set_xlabel('Height', size=15)
ax[1].set_xlabel('Height', size=15)

# Y축 이름 설정
ax[0].set_ylabel('Frequency', size=15)  # 빈도(개수)
ax[1].set_ylabel('Density', size=15)    # 확률밀도

plt.show()  # 완성된 그래프 화면에 출력


# density 차이
# density=False → “각 구간에 몇 명?” (개수)
# density=True → “전체 분포를 확률밀도 형태로 표현”
# ※ density=True를 단순히 비율(%)이라고 설명하면 정확하지 않음. 확률밀도라고 하는 게 맞음
````

> ## 2) Multiple Plots

````
# 4가지 종류의 그래프를 2행 2열로 한 화면에 시각화
# subplot을 이용해서 Bar / Box / Scatter / Pie 그래프를 한 Figure에 배치하는 종합 실습

# 2행 2열의 그래프 영역을 생성
fig, ax = plt.subplots(
    nrows=2,             # 행의 개수 → 2행
    ncols=2,             # 열의 개수 → 2열
    figsize=(15, 10)     # 전체 Figure 크기
)

# ① 왼쪽 위 → Height와 Weight를 막대 그래프로 비교
ax[0, 0].bar(
    DF.index, DF['Height'],
    width=0.3,
    label='Height'
)
ax[0, 0].bar(
    DF.index + 0.3, DF['Weight'],
    width=0.3,
    label='Weight'
)

# ② 오른쪽 위 → 혈액형별 Height 분포를 박스플롯으로 비교
ax[0, 1].boxplot(
    [
        DF.loc[DF['BloodType'] == 'A', 'Height'],   # A형의 키
        DF.loc[DF['BloodType'] == 'B', 'Height'],   # B형의 키
        DF.loc[DF['BloodType'] == 'O', 'Height'],   # O형의 키
        DF.loc[DF['BloodType'] == 'AB', 'Height']   # AB형의 키
    ],
    labels=['A', 'B', 'O', 'AB'],  # 혈액형 이름
    patch_artist=True,             # 박스 내부에 색상을 채울 수 있도록 설정
    showmeans=True                 # 평균값 표시
)

# ③ 왼쪽 아래 → Height와 Weight의 관계를 산점도로 표시
ax[1, 0].scatter(
    DF.Height, DF.Weight,
    marker='*',                    # 별 모양의 점
    s=100,                         # 점의 크기
    c='#d62728'                    # 점의 색상
)

# ④ 오른쪽 아래 → 혈액형별 비율을 원형 그래프로 표시
ax[1, 1].pie(
    DF.BloodType.value_counts(),              # 혈액형별 인원수
    labels=DF.BloodType.value_counts().index, # 혈액형 이름
    autopct='%.1f%%',                         # 비율을 %로 표시
    startangle=90,                            # 90도 위치에서 시작
    explode=[0.0, 0.0, 0.0, 0.05]            # 네 번째 조각을 조금 떨어뜨려 강조
)

# 막대 그래프의 범례(이름표) 표시
ax[0, 0].legend()

# 각 그래프의 제목 설정
ax[0, 0].set_title('Bar Plot')
ax[0, 1].set_title('Box Plot')
ax[1, 0].set_title('Scatter Plot')
ax[1, 1].set_title('Pie Plot')

# 각 그래프의 X축 이름 설정
ax[0, 0].set_xlabel('Height & Weight')
ax[0, 1].set_xlabel('Blood Type')
ax[1, 0].set_xlabel('Height')
ax[1, 1].set_xlabel('')

# 각 그래프의 Y축 이름 설정
ax[0, 0].set_ylabel('Frequency')
ax[0, 1].set_ylabel('Height')
ax[1, 0].set_ylabel('Weight')
ax[1, 1].set_ylabel('')

plt.show()  # 완성된 4개의 그래프를 화면에 출력
````

# seaborn Package

Matplotlib이 기본 도구이고, Seaborn은 Matplotlib을 더 편하고 예쁘게 사용할 수 있게 만든 라이브러리
| 구분     | Matplotlib     | Seaborn                  |
| ------ | -------------- | ------------------------ |
| 발음     | 맷플롯립           | 시본                       |
| 특징     | 기본적인 그래프 라이브러리 | Matplotlib 기반의 고수준 라이브러리 |
| 사용     | 세밀하게 직접 설정     | 적은 코드로 보기 좋은 그래프         |
| 데이터 분석 | 가능             | **통계·데이터 분석 시각화에 특히 편리** |
| 대표 코드  | `plt.plot()`   | `sns.lineplot()`         |
| 관계     | 기반 라이브러리       | **내부적으로 Matplotlib 사용**  |
```
Matplotlib = 그래프를 직접 만드는 기본 도구
Seaborn = Matplotlib을 기반으로 데이터 분석 그래프를 더 쉽고 보기 좋게 만드는 도구

# Matplotlib
plt.plot(DF.index, DF.Height)

# Seaborn
sns.lineplot(x=DF.index, y=DF.Height)
상관관계, 분포, 그룹별 비교 같은 데이터 분석을 할 때 Seaborn이 편함

````
# 그래프 출력을 위한 matplotlib 불러오기
import seaborn as sns
import matplotlib.pyplot as plt
````

# I. 선 그래프

* https://seaborn.pydata.org/generated/seaborn.lineplot.html#seaborn.lineplot

````
# 그래프 크기를 설정하고 새로운 그래프 영역 생성
plt.figure(figsize=(10, 7))

# Seaborn을 이용하여 Height(키) 데이터를 선 그래프로 시각화
sns.lineplot(
    x=DF.index,        # X축 → 데이터의 인덱스
    y=DF.Height,       # Y축 → Height(키) 데이터
    linewidth=1,       # 선의 두께
    color='r',         # 선 색상 → 빨간색(red)
    marker='>',        # 각 데이터 위치를 > 모양으로 표시
    linestyle='--'     # 선 모양 → 점선
)

plt.title('Line Graph', size=30)  # 그래프 제목
plt.xlabel('Index', size=20)      # X축 이름
plt.ylabel('Height', size=20)     # Y축 이름
plt.grid(True)                    # 격자 표시

plt.show()                        # 완성된 그래프 화면에 출력
````

# VII. Histograms

````
# Seaborn을 이용하여 1행 2열의 그래프 영역 생성
fig, ax = plt.subplots(
    nrows=1,            # 행의 개수 → 1행
    ncols=2,            # 열의 개수 → 2열
    figsize=(15, 5)     # 전체 Figure 크기
)

# ① 왼쪽 → Height(키)의 히스토그램 + KDE 밀도곡선
sns.histplot(
    data=DF,            # 그래프에 사용할 DataFrame
    x='Height',         # X축 → Height(키)
    bins=5,             # 데이터를 5개의 구간으로 나눔
    alpha=0.2,          # 막대의 투명도
    kde=True,           # KDE(밀도곡선)를 함께 표시
    ax=ax[0]            # 첫 번째 그래프 영역에 출력
)

# ② 오른쪽 → Height(키)의 KDE 밀도곡선
sns.kdeplot(
    data=DF,            # 그래프에 사용할 DataFrame
    x='Height',         # X축 → Height(키)
    color='royalblue',  # 곡선 색상
    ax=ax[1]            # 두 번째 그래프 영역에 출력
)

# 실제 Height 데이터의 위치를 작은 선으로 표시
sns.rugplot(
    data=DF,
    x='Height',
    ax=ax[1]
)

# 각 그래프의 제목 설정
ax[0].set_title('Histogram + KDE', size=20)
ax[1].set_title('KDE + Rug Plot', size=20)

# X축 이름 설정
ax[0].set_xlabel('Height', size=15)
ax[1].set_xlabel('Height', size=15)

# Y축 이름 설정
ax[0].set_ylabel('Frequency', size=15)
ax[1].set_ylabel('Density', size=15)

plt.show()

# 왼쪽: histplot() → 구간별 개수 + 분포 곡선
# 오른쪽: kdeplot() → 분포를 부드러운 곡선으로, rugplot() → 실제 데이터 위치 표시
````

> ## 2) Multiple Plots

````
# Seaborn을 이용하여 2행 2열의 그래프 영역 생성
fig, ax = plt.subplots(
    nrows=2,             # 행의 개수 → 2행
    ncols=2,             # 열의 개수 → 2열
    figsize=(15, 10)     # 전체 Figure 크기
)

# ① 왼쪽 위 → 학년별 평균 나이를 성별로 구분하여 막대 그래프로 표시
sns.barplot(
    data=DF,
    x='Grade',           # X축 → 학년
    y='Age',             # Y축 → 나이의 평균
    hue='Gender',        # 성별에 따라 막대를 구분
    errorbar=None,       # 오차막대를 표시하지 않음
    palette='deep',
    ax=ax[0, 0]          # 1행 1열에 출력
)

# ② 오른쪽 위 → 몸무게의 분포를 히스토그램으로 표시
sns.histplot(
    data=DF,
    x='Weight',          # X축 → 몸무게
    bins=6,              # 데이터를 6개의 구간으로 나눔
    alpha=0.3,           # 막대의 투명도
    ax=ax[0, 1]          # 1행 2열에 출력
)

# ③ 왼쪽 아래 → 혈액형별 키의 분포를 박스플롯으로 비교
sns.boxplot(
    data=DF,
    x='BloodType',       # X축 → 혈액형
    y='Height',          # Y축 → 키
    order=['A', 'B', 'O', 'AB'],  # 혈액형 표시 순서
    palette='deep',
    ax=ax[1, 0]          # 2행 1열에 출력
)

# ④ 오른쪽 아래 → 키와 몸무게의 관계를 산점도로 표시
sns.scatterplot(
    data=DF,
    x='Height',          # X축 → 키
    y='Weight',          # Y축 → 몸무게
    hue='Grade',         # 학년에 따라 점의 색상을 구분
    style='BloodType',   # 혈액형에 따라 점의 모양을 구분
    s=50,                # 점의 크기
    ax=ax[1, 1]          # 2행 2열에 출력
)

# 첫 번째 그래프의 범례 위치와 제목 설정
ax[0, 0].legend(
    loc='upper left',    # 범례 위치 → 왼쪽 위
    title='Gender'       # 범례 제목
)

# 각 그래프의 제목 설정
ax[0, 0].set_title('Bar Plot')
ax[0, 1].set_title('Histogram')
ax[1, 0].set_title('Box Plot')
ax[1, 1].set_title('Scatter Plot')

# X축 이름 설정
ax[0, 0].set_xlabel('Grade')
ax[0, 1].set_xlabel('Weight')
ax[1, 0].set_xlabel('Blood Type')
ax[1, 1].set_xlabel('Height')

# Y축 이름 설정
ax[0, 0].set_ylabel('Age Mean')
ax[0, 1].set_ylabel('Frequency')
ax[1, 0].set_ylabel('Height')
ax[1, 1].set_ylabel('Weight')

plt.show()

# hue = 색상으로 그룹 구분
# style = 점 모양으로 그룹 구분
# ax = 여러 그래프 중 어디에 그릴지 지정
# errorbar=None = 평균 주변의 오차막대는 표시하지 않음
````
