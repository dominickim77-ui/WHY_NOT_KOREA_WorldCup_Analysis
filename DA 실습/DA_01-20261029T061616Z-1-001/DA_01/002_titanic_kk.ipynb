# 타이타닉 승객 생존 데이터 분석 보고서

## 1. 분석 목적

타이타닉 승객 데이터를 이용하여 **어떤 요인이 승객의 생존 여부와 관련이 있는지** 분석한다.

특히 다음 항목을 중심으로 비교한다.

* 전체 승객의 생존 및 사망 현황
* 성별에 따른 생존 차이
* 객실 등급에 따른 생존 차이
* 어떤 속성이 생존 여부와 더 큰 관련성을 보이는지 확인



## 2. 데이터 개요

분석에 사용한 데이터는 **총 156명의 타이타닉 승객 데이터**이다.

`SURVIVED` 컬럼은 승객의 생존 여부를 나타낸다.

| 값 | 의미 |
| - | -- |
| 0 | 사망 |
| 1 | 생존 |

전체 156명 중 **102명이 사망하고 54명이 생존**하였다.

| 구분 |   인원 |    비율 |
| -- | ---: | ----: |
| 사망 | 102명 | 65.4% |
| 생존 |  54명 | 34.6% |
| 전체 | 156명 |  100% |

따라서 전체 승객 중 약 **65%가 사망**한 것으로 나타났다.



## 3. 성별에 따른 생존 분석

성별에 따른 생존 여부를 비교한 결과 다음과 같다.

| 성별 |   전체 |  사망 |  생존 |   생존율 |
| -- | ---: | --: | --: | ----: |
| 남성 | 100명 | 86명 | 14명 | 14.0% |
| 여성 |  56명 | 16명 | 40명 | 71.4% |

남성은 100명 중 **86명이 사망**했으며 생존율은 **14.0%**에 불과했다.

반면 여성은 56명 중 **40명이 생존**하여 생존율이 **71.4%**로 나타났다.

 **성별에 따라 생존율에 매우 큰 차이가 나타난다.**



## 4. 객실 등급에 따른 생존 분석

`PCLASS`는 승객의 객실 등급을 의미한다.

* 1 : 1등급
* 2 : 2등급
* 3 : 3등급

| 객실 등급 |  전체 |  사망 |  생존 |   생존율 |
| ----- | --: | --: | --: | ----: |
| 1등급   | 30명 | 18명 | 12명 | 40.0% |
| 2등급   | 30명 | 16명 | 14명 | 46.7% |
| 3등급   | 96명 | 68명 | 28명 | 29.2% |

3등급 승객의 사망자는 **68명으로 가장 많았고**, 생존율도 **29.2%로 가장 낮았다.**

하지만 3등급 승객 자체가 96명으로 다른 등급보다 많기 때문에 **사망자 수만으로 판단하기보다는 생존율을 함께 비교해야 한다.**



## 5. 분석 결과

이번 데이터에서는 **성별과 객실 등급 모두 생존 여부와 관련된 차이**가 나타났다.

특히 성별에서는 남성의 생존율이 **14.0%**, 여성의 생존율이 **71.4%**로 큰 차이를 보였다.

객실 등급에서는 3등급 승객의 생존율이 **29.2%로 가장 낮게 나타났다.**

따라서 이 데이터만 놓고 보면 **객실 등급보다 성별에서 생존율 차이가 더 뚜렷하게 나타난다.**

> **결론 : 타이타닉 승객의 생존 여부에는 여러 요인이 관련되어 있지만, 이 데이터에서는 특히 성별이 생존 여부와 강한 관련성을 보였다.**

※ 단, 이것은 현재 156명의 표본에서 관찰된 결과이며, 이것만으로 성별이 생존을 직접적으로 결정했다고 단정할 수는 없다.
