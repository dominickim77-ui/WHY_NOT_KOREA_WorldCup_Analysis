##  DataFrame
- **DataFrame**은 행(row)과 열(column)로 이루어진 표 형태의 데이터

- `head()`는 앞부분 데이터를 확인할 때 사용험
- `read_csv()`는 CSV 파일을 DataFrame으로 읽어오는 함수이다.

```python
### 2차원 리스트를 데이터프레임으로 변환
# 파이썬의 리스트(list) 는 자바스크립트의 배열(array) 과 비슷, mutable(변경 가능), 튜플(tuple)이 불변(immutable)
# 순서가 있는 자료형 → 인덱스로 접근 가능 ([0], [1], …)
# 여러 타입 혼합 저장 가능 → 숫자, 문자열, 객체 등을 함께 저장 가능
# 변경 가능(mutable) → 요소를 추가, 수정, 삭제 가능
score=[['정약용', 70, 80, 100],
       ['이순신', 100, 80, 70],
       ['이율곡', 90, 95, 90]]
print(type(score))
score
```

```python
score=[
   ['정약용', 70, 80, 100],
   ['이순신', 100, 80, 70],
   ['이율곡', 90, 95, 90]
]
print(type(score))
score
```

```python
score_df = pd.DataFrame(score)  # 2차원 리스트 → DataFrame 변환
print(type(score_df))           # 자료형 확인
score_df                        # Jupyter에서 자동 출력

# Pandas의 DataFrame은 엑셀 표(스프레드시트)와 거의 같은 구조
# DataFrame = 표
# 행(Row) → 데이터 한 건(예: 한 사람의 점수, 한 상품의 정보)
# 열(Column) → 데이터 속성(예: 이름, 국어 점수, 수학 점수)
# 인덱스(Index) → 행 번호(엑셀의 맨 왼쪽 행 번호처럼), 고유 ID 역할 가능
# 컬럼명(Columns) → 열 이름(엑셀의 맨 위 열 제목처럼)
```

```python
score_df = pd.DataFrame(score)
print(type(score_df))
score_df
```

## 데이터 상태 확인
- `info()` : 열 이름, 데이터 타입, 결측치 여부 등을 확인한다.
- `describe()` : **숫자형 열 중심으로** 평균, 표준편차, 최솟값, 최댓값, 사분위수를 보여준다.
- 분석을 시작하면 모델부터 만드는 것이 아니라 **데이터 상태를 먼저 확인**하는 습관이 중요하다.


```python
# [수업 포인트] info()로 열 이름, 자료형, 결측치 여부를 확인
score_df.info()  # DataFrame 정보 요약
# .info() 메서드는 행/열 개수, 컬럼명, 결측치 여부, 자료형, 메모리 사용량을 보여준다.
# 데이터 구조를 빠르게 파악하고 결측치 확인에 유용하다.
# Pandas의 자료형(dtype)은 NumPy 자료형을 기반
# NumPy에는 파이썬의 문자열 전용 타입이 따로 없고, 모든 문자열을 파이썬 객체로 취급
```

```python
# <class 'pandas.core.frame.DataFrame'>   이 객체가 Pandas DataFrame이라는 뜻
# RangeIndex: 3 entries, 0 to 2           행 인덱스 범위(0~2), 총 3개의 행
# Data columns (total 4 columns)         전체 열 개수(4개)
#       #    /    Column   /    Non-Null Count    /    Dtype
#    열 번호 /   열 이름  /  결측치가 아닌 값의 개수 /  Dtype : 자료형(int64, float64, object 등)
# dtypes: int64(3), object(1)   숫자형 열이 3개, 문자열(object) 열이 1개
# memory usage   이 DataFrame이 차지하는 메모리 양
```

```python
n=pd.DataFrame({'col_str1':['1', '2', '3', '4'],
                'col_str2':['5', '6', '7', '8']})
print(n)
n.dtypes   # 각 열의 자료형 확인,   '1' 은 문자열-> 넘파이에서 객체로 인식
```

| 방식        | 기준           | 특징      |
| --------- | ------------ | ------- |
| 리스트 안 리스트 | Row(행) 중심    | 한 줄씩 입력 |
| 딕셔너리      | Column(열) 중심 | 열 단위 입력 |


| 코드 방식           | 데이터 방향        |
| --------------- | ------------- |
| `[[...],[...]]` | 가로(row) 입력    |
| `{컬럼:[...]}`    | 세로(column) 입력 |


| 방식        | 의미              |
| --------- | --------------- |
| 리스트 안 리스트 | 행(row) 기준 입력    |
| 딕셔너리      | 열(column) 기준 입력 |

```python
n_df=n.astype({'col_str1':int, 'col_str2':float})
print(n_df)
n_df.dtypes
```

| 구분        | 역할       | 위치 | 예시           |

```python
score_df.drop(['성명'], axis=1 , inplace=True)  # "성명"항목 삭제
score_df
## inplace=True : 결과를 변수 score_df에 반영

# drop() : 행 또는 열을 제거하는 메서드, 삭제 방향 지정: axis=0(행), axis=1(열)
# ['성명'] : 삭제할 컬럼 이름(리스트 형태로 지정)
# axis=1 : 열(Column) 기준으로 삭제
# (axis=0이면 행(Row) 삭제)
# inplace=True : 새로운 객체를 반환하지 않고, 현재 DataFrame(score_df)에 바로 반영
# inplace는 제자리에, 원래 위치라는 뜻
# inplace=True → 원본 DataFrame/Series를 직접 수정
# inplace=False(기본값) → 원본은 그대로 두고, 변경된 복사본을 반환
```

| 값        | 의미           |
| -------- | ------------ |
| `axis=0` | 행(row) 방향    |
| `axis=1` | 열(column) 방향 |


```python
score_df.drop(['이순신'], axis=0,inplace=True)  # 인덱스 '이순신' 행 삭제 (원본에 반영)
```
