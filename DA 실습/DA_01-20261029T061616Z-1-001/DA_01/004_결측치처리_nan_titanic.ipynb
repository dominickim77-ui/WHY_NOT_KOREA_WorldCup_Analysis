## 결측치(NaN)
- 결측치는 값이 비어 있는 상태이다.
- `isnull().sum()`으로 열별 결측치 개수를 확인할 수 있다.
- `dropna()`는 결측치가 있는 데이터를 제거하고,
- `fillna()`는 평균·중앙값 등 다른 값으로 채운다.
- 어떤 방법이 무조건 정답은 아니고 **데이터의 의미를 보고 결정**해야 한다.


````python
# [수업 포인트] 결측치(NaN)가 어디에 얼마나 있는지 확인
### 결측치 확인-2
air.isnull() # 값이 비어있으면(True), 값이 있으면(False)
````

````python
# [수업 포인트] 결측치(NaN)가 어디에 얼마나 있는지 확인
### 결측치 확인-3
air.isnull().sum()  # 열(column)별 결측치 개수를 세어준다.
````

#### 결측치 처리_결측치 제거

````python
# [수업 포인트] 결측치가 있는 행/열을 제거하는 전처리
air_d = air.dropna(axis=0)    ## 결측치가 있는 행을 삭제  , na자리를 버림, na의 행을 버림
air_d

# 결측치(NaN)가 하나라도 있는 행을 전부 삭제해서 air_d를 만듬
# axis=0: "행 기준으로"라는 뜻(사실 기본값이라 생략해도 됨)
# 원본 보존: air는 그대로, air_d에 결과가 담김
````

| 옵션       | 의미   |
| -------- | ---- |
| `axis=0` | 행 기준 |
| `axis=1` | 열 기준 |


#### 결측치 처리 - 평균값으로 대체

````python
# 결측치 평균대체를 위해 평균 확인, 숫자항목만 평균으로 구해라
mean_air = air.mean(numeric_only=True)
mean_air

# 숫자형 열들만 평균을 구해서(결측치는 자동 제외) Series로 반환
# air에서 숫자형 컬럼만 골라 평균을 계산
# numeric_only=True? 문자/날짜 컬럼은 빼고 숫자만 계산 (누메릭-수치)
# 평균 계산 시 NaN은 자동으로 제외 (평균 계산 시 결측치는 제외)
````

| 코드                  | 의미         |
| ------------------- | ---------- |
| `mean()`            | 평균 계산      |
| `numeric_only=True` | 숫자 데이터만 사용 |


| 컬럼     | 타입           |
| ------ | ------------ |
| MMDDHH | object(문자형)  |
| PM10   | float64(숫자형) |
| PM25   | float64      |
| O3     | float64      |


````python
print(air)
````

````python
# [수업 포인트] 결측치를 다른 값으로 채우는 전처리
#  평균값으로 채워라
air_m = air.fillna(mean_air)  # na 값으로 채움
air_m
# fillna가 결측치만 채우고 나머지는 그대로
# NaN(Not a Number)라는 값은 결측치를 의미, 즉, 데이터가 비어있거나(null) 숫자가 아닌 상태
````

#### 결측치 처리 - 중앙값으로 대체

````python
air_md = air.median(numeric_only=True)
air_md
# 숫자형 열들만 중앙값(미디언)을 계산해 Series로 반환
# air의 숫자형 컬럼만 골라 중앙값을 구한다.
# 결측치 처리는 중앙값 계산 시 NaN은 자동 제외
# numeric_only=True 는 숫자형 열만 골라서 계산하라는 뜻
# 문자열(object), 범주형(category), 날짜/시간(datetime) 같은 비숫자 열은 무시
# 보통 mean(), median(), sum(), std() 같은 집계 함수에서 사용
# bool 열은 숫자로 취급되어 포함됩니다(True=1, False=0)
````

````python
# [수업 포인트] 결측치를 다른 값으로 채우는 전처리
air_md = air.fillna(air_md)   # 중앙값으로 결측 대치
air_md
# 중앙값으로 결측 대치한 DataFrame로 덮어씀
````

##  groupby / agg
- `groupby()`는 같은 기준끼리 데이터를 묶는 기능이다.
- 예: 반별, 성별, 고객별, 월별로 묶기
- `agg()`는 묶은 데이터에 평균·합계·개수·최댓값 등을 한 번에 계산할 때 사용한다.
- 실무 데이터 분석에서는 **'묶고 → 요약하기'**가 매우 자주 사용된다.


````python
# [수업 포인트] 같은 기준의 데이터를 그룹으로 묶어서 요약
titanic_df.groupby('embark_town').size()   # size  카운트

# titanic_df.groupby('embark_town').size()는 탑승한 항구별 승객 수(숫자)를 구하는 코드
# .groupby()는 데이터를 특정 기준(그룹)으로 묶어서 연산을 수행하는 기능
# 엑셀의 피벗 테이블이나 그룹별 합계/평균 같은 기능
# embark_town 값별로 데이터 개수를 세는 코드
# 644는 Southampton에서 탄 승객 수
# embark_town(탑승 항구) 값은 타이타닉 데이터셋에서 원래 3개 도시만 존재
````

| 코드                       | 의미         |

결측치(NaN)를 처리하는 이유

결측치란?
데이터를 수집하는 과정에서 값이 없거나 측정되지 않아 비어 있는 데이터를 의미한다.
```
예)
나이 : 25
성별 : 남자
키   : NaN  ← 값이 없음(결측치)
```
결측치를 처리하는 이유

데이터가 비어 있으면 평균 계산, 통계 분석, 머신러닝 학습 등에 문제가 발생하거나 분석 결과가 부정확해질 수 있기 때문이다.

따라서 분석하기 전에 결측치를 확인하고 상황에 따라 처리한다.

##### 삭제 → 결측치가 있는 행이나 열을 제거
##### 대체 → 평균값, 중앙값, 최빈값 등으로 채움



#### 결측치(NaN) = 0이 아니라, 데이터가 없다는 뜻
#### → 분석에 방해가 될 수 있으므로 전처리 과정에서 적절하게 처리한다.
