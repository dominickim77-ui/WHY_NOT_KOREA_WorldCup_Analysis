##  알아두면 좋은 개념: groupby / agg
- `groupby()`는 같은 기준끼리 데이터를 묶는 기능이다.
- 예: 반별, 성별, 고객별, 월별로 묶기
- `agg()`는 묶은 데이터에 평균·합계·개수·최댓값 등을 한 번에 계산할 때 사용한다.
- 실무 데이터 분석에서는 **'묶고 → 요약하기'**가 매우 자주 사용된다.


```python
# [수업 포인트] 같은 기준의 데이터를 그룹으로 묶어서 요약
exam.groupby('nclass')[['math', 'english','science']].mean()   # 반별로 점수,  nclass 기준으로   .mean() 평균
```

| 코드                               | 의미       |
| -------------------------------- | -------- |
| groupby('nclass')                | 반별로 묶기   |
| [['math', 'english', 'science']] | 여러 과목 선택 |
| mean()                           | 평균 계산    |


| 형태                  | 의미      |
| ------------------- | ------- |
| df['컬럼명']         | 컬럼 1개   |
| df[['컬럼1','컬럼2']] | 컬럼 여러 개 |


| 문법                     | 이유                |
| ---------------------- | ----------------- |
| ['math']             | 컬럼 1개             |
| [['math','english']] | 여러 컬럼             |
| 바깥 []                  | DataFrame에서 컬럼 선택 |
| 안쪽 []                  | 컬럼 이름들을 리스트로 전달   |
##### [ ['컬럼1','컬럼2'] ] 컬럼 여러 개를 선택하겠다

```python
# [수업 포인트] 같은 기준의 데이터를 그룹으로 묶어서 요약
# [수업 포인트] 묶은 데이터에 평균/합계/개수 등 여러 통계를 적용
exam.groupby('nclass').agg(mean_math   = ('math', 'mean'),    # 수학 평균
                         sum_math    = ('math', 'sum'),       # 수학 점수 합계
                         median_math = ('math', 'median'),    # math 컬럼 중앙값
                         n = ('nclass', 'count'))             # 학생 수
# groupby('nclass') → 반(nclass)별로 데이터를 묶음
# .agg() → 묶은 데이터에 대해 여러 계산을 동시에 적용
# median_math →
# n → 해당 반의 학생 수(행 개수)
```

| 코드                  | 의미    |
| ------------------- | ----- |
| ('math', 'mean')  | 수학 평균 |
| ('math', 'sum')   | 수학 합계 |
| ('math', 'max')   | 최고점   |
| ('math', 'min')   | 최저점   |
| ('math', 'count') | 개수    |


##### aggregation(애그리게이션) : 여러 데이터를 모아서 하나로 계산하거나 요약하는 것
| aggregation 종류 | 의미            |
| -------------- | ------------- |
| 평균(mean)       | 여러 값을 평균으로 요약 |
| 합계(sum)        | 여러 값을 더해서 요약  |
| 개수(count)      | 몇 개인지 세기      |
| 최대값(max)       | 가장 큰 값 찾기     |
| 최소값(min)       | 가장 작은 값 찾기    |

### 데이터 합치기 _ merge(), concat()

```python
# 중간고사 데이터 만들기

test1 = pd.DataFrame({'id'      : [1, 2, 3, 4, 5, 7],
                      'midterm' : [60, 80, 70, 90, 85, 20]})

# 기말고사 데이터 만들기
test2 = pd.DataFrame({'id'    : [1, 2, 3, 4, 5, 6],
                      'final' : [70, 83, 65, 95, 80, 10]})
```

## 알아두면 좋은 개념: 데이터 결합
- 실제 업무 데이터는 한 파일에 모든 정보가 들어 있지 않은 경우가 많다.
- 그래서 고객정보 + 이용이력 + 상품정보처럼 여러 표를 연결해야 한다.
- `merge()`는 공통 열(Key)을 기준으로 두 DataFrame을 연결하는 함수이다.
- 데이터베이스의 **JOIN**과 거의 같은 개념으로 생각하면 된다.


```python
# [수업 포인트] 공통 열(Key)을 기준으로 여러 DataFrame을 연결
# id 기준으로 merge
total = pd.merge( test1, test2, on='id', how='left')  ##  how='outer'로 하면, 양쪽 모든 데이터
total

#  n='id' → id 컬럼을 기준으로 두 데이터를 결합
# how='left' → test1(왼쪽 데이터프레임)에 있는 모든 id를 기준으로 합침
# test1에는 있고 test2에는 없는 id=7 → final 값은 NaN
# test2에만 있는 id=6 → 결과에 포함되지 않음
```

```python
# [수업 포인트] 공통 열(Key)을 기준으로 여러 DataFrame을 연결
# id 기준으로 merge
total = pd.merge( test1, test2, on='id', how='outer')  ##  how='outer'로 하면, 양쪽 모든 데이터
total

# how 옵션 차이
# 'left'	왼쪽(test1) 기준으로 병합	왼쪽 데이터 전부 + 오른쪽 매칭
# 'right'	오른쪽(test2) 기준으로 병합	오른쪽 데이터 전부 + 왼쪽 매칭
# 'inner'	양쪽 모두에 있는 id만 병합	교집합
# 'outer'	양쪽 모든 id 병합	합집합(없는 값은 NaN)
```

```python
group_a = pd.DataFrame({'id' : [1, 2, 3, 4, 5],
                        'test':[60, 80, 70, 90, 85] })

group_b = pd.DataFrame({'id' : [6, 7, 8, 9, 10],
                        'test':[70, 83, 65, 95, 80] })

print(group_a)

print(group_b)
```

```python
# concat() 이용하여 합치기
group_all = pd.concat([group_a, group_b], ignore_index=True)
group_all

# 두 DataFrame을 세로 방향으로 합치기
# ignore_index=True → 기존 인덱스를 무시하고 0부터 재배열 (위에서부터 번호를 새로 다시 부여함)
```

| 옵션                   | 의미           |
| -------------------- | ------------ |
| `ignore_index=False` | 기존 번호 유지     |
| `ignore_index=True`  | 번호 새로 다시 만들기 |


### loc 와  iloc

```python
#loc  loaction      iloc  interger 행과 열을 숫자로 표현해서 찾게
```

```python
# 예제 데이터프레임
# 딕셔너리 형태
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'Score': [85, 90, 95]
}
df = pd.DataFrame(data, index=['A', 'B', 'C'])  # DataFrame 생성, 인덱스를 A, B, C로 지정
df
```

#### loc

```python
df.loc[['A']]  # 행 'A' 조회   loc는 행이나 문자열로
# 행 'A' 조회 (리스트 형태로 전달 → DataFrame 반환)

# .loc[행인덱스, 열인덱스]
# - 행인덱스: 문자열(라벨) 또는 조건식 가능
# - 열인덱스: 열 이름 지정
# - 리스트로 지정하면 DataFrame, 단일 값이면 Series 반환
```

```python
df.loc['A':'B', 'Name']  # 행 'A'부터 'B'까지 조회 (끝값 포함)   1-3  이면 1,2
# loc는 라벨(이름) 기반이므로 'A':'B' 라고 하면 'A'부터 'B'까지 끝값(B) 포함해서 가져온다.
# iloc는 정수 위치(index) 기반이므로 1:3 라고 하면 1, 2번 인덱스까지만 가져오고 끝값(3)은 제외한다.
```

```python
df.loc[df['Age'] > 28]  # Age가 28보다 큰 행 조회

# df['Age'] > 28 → True/False 값을 가진 불리언 시리즈 생성
# A    False
# B     True
# C     True
# Name: Age, dtype: bool
# .loc[조건] → True인 행만 반환
# df는 pd.DataFrame(...)으로 만든 데이터프레임(DataFrame) 객체
# df.라고 쓰면 그 DataFrame이 가지고 있는 속성과 메서드를 쓸 수 있다.
```

### iloc

```python
df.iloc[0]  # 첫 번째 행 조회
```

```python
df.iloc[0:2]  # 0번부터 1번까지 (끝값 포함 X)   iloc 는 인티져값,  loc는 행이나 문자열로
```

| 코드             | 결과 타입     | 출력 모습 |
| -------------- | --------- | ----- |
| `df.iloc[0]`   | Series    | 세로    |
| `df.iloc[0:2]` | DataFrame | 가로 표  |


##### iloc[0] → 한 행을 Series 형태로 가져와서 세로로 보임
##### iloc[0:2] → 여러 행을 DataFrame 형태로 가져와서 가로 표 형태로 보임

정리하면:
| 가져오는 개수 | 타입        |
| ------- | --------- |
| 한 줄만    | Series    |
| 여러 줄    | DataFrame |
