## 회귀분석
- 회귀는 입력값(X)을 이용해 **숫자형 결과(y)**를 예측하는 방법이다.
- 기본 모양은 `y = ax + b`로 생각하면 이해하기 쉽다.
- `coef`는 X가 1 증가할 때 y가 얼마나 변하는지를 나타낸다.
- 학습용(train)과 테스트용(test)을 나누는 이유는 **처음 보는 데이터에서도 잘 맞는지 확인**하기 위해서이다.


````python
# 입력값과 결과값 사이의 관계를 직선/회귀식으로 학습
# 여러 자동차 특성을 이용하여 연비(mpg)를 예측하는 다중선형회귀

# 종속변수(y) : mpg → 예측할 값(연비)
# 독립변수(x) : cylinders, displacement, weight, acceleration, model_year
# 여러 독립변수는 + 로 연결


# 1. 다중선형회귀 모델 생성 (연비 = 실린더수 + 배기량 + 무게 + 가속도 + 연식)
                               # 종속변수    ~ 독립변수
                               # y  ,      x1 ,       x2   ,       x3 ,      x4         + x5
model_mlm = smf.ols(formula = 'mpg ~ cylinders + displacement +   weight + acceleration +   model_year', data = data_df)

result_mlm = model_mlm.fit() #  2. 모델 학습 → 데이터를 이용하여 회귀계수 계산

result_mlm.summary() # 3. 결과 확인 → R²(설명력), coef(계수), p-value 등
````

| 꼭 볼 값                  |          결과 | 쉽게 해석                                 |
| ---------------------- | ----------: | ------------------------------------- |
| **R-squared**          |   **0.809** | 5개 변수로 연비의 변화를 **약 80.9% 설명**         |
| `cylinders` p-value    |       0.432 | 0.05보다 큼 → 연비와 **뚜렷한 관계가 있다고 보기 어려움** |
| `displacement` p-value |       0.310 | 0.05보다 큼 → 연비와 **뚜렷한 관계가 있다고 보기 어려움** |

| `acceleration` p-value |       0.306 | 0.05보다 큼 → 연비와 **뚜렷한 관계가 있다고 보기 어려움** |
| `model_year` coef      |  **0.7553** | 연식이 1 증가하면 예측 연비는 약 **0.76 증가**       |
| `model_year` p-value   |   **0.000** | 0.05보다 작음 → 연비와 **의미 있는 관계가 있음**      |

이 모델은 연비 변화를 약 80.9% 설명하며, 5개 변수 중 weight(차량 무게)와 model_year(연식)이 연비와 뚜렷한 관계를 보였다.

특히  차가 무거울수록 연비는 낮아지고, 연식이 최근일수록 연비는 높아지는 방향으로 나타났다.
단, 이 결과는 다른 변수들을 같이 고려한 회귀분석 결과이다.

````python
# [수업 포인트] 입력값과 결과값 사이의 관계를 직선/회귀식으로 학습
# # 다중회귀 모델 생성 (연비 = 배기량 + 무게 + 가속도 + 연식)
model_mlm = smf.ols(formula = 'mpg ~ displacement +   weight + acceleration +   model_year', data = data_df)

result_mlm = model_mlm.fit() # # 모델 학습 (회귀식 계산)

result_mlm.summary()  # # 결과 출력 (계수, p값, 설명력 확인)
````

````python
y = data_df['mpg']     ## 종속변수, y = 맞춰야 할 정답
x = data_df.drop(['mpg'], axis=1, inplace=False)   # 독립변수들,x = 정답을 맞추기 위한 재료(입력값)
````

````python

"""
자동차 연비 예측이라고 하면
mpg → 연비 (정답)  y
나머지 (무게, 마력 등) → 입력 x
"""
````

````python
# [수업 포인트] 학습용 데이터와 테스트용 데이터를 나눠 모델의 일반화 성능을 확인
from sklearn.model_selection import train_test_split

# 훈련용 데이터(70%)와 평가용 데이터(30%)로 분할
# random_state=0 다시 실행했을때 랜덤으로 ７０개／３０개 같은 데이터가지고 진행할수 있게  random_state=４
# 전체 데이터의 30%를 시험용으로 사용
x_train, x_test, y_train, y_test = train_test_split(x, y,  test_size=0.3,  random_state=0) # random_state=0 랜덤하게 나누되, 매번 똑같이 나누도록 고정

# 전체 데이터 개수
print("전체 데이터 :", len(x))

# 학습용 데이터 개수
print("X_train :", len(x_train))
print("y_train :", len(y_train))

# 시험용 데이터 개수
print("X_test :", len(x_test))
print("y_test :", len(y_test))
````

```

실제로 train_test_split()은 반올림 처리 때문에 이 데이터에서는 학습 278개 / 테스트 120개로 나눈다.
```

| y_train | 학습용 정답 |
| x_test  | 시험용 입력 |
| y_test  | 시험용 정답 |
##### X_train → 문제집
##### y_train → 정답지
##### X_test → 시험 문제
##### y_test → 시험 정답

````python
x_train  # 모델을 공부시키는 입력 데이터, 연비(mpg)를 예측하기 위한 입력값들
# X → 입력
# train → 학습용
````

````python
# 입력값(X)과 결과값(y)의 관계를 학습하여 연비를 예측

# 1. 선형회귀 모델 생성
lr = LinearRegression()
# → 선형회귀 모델을 사용할 준비


# 2. 모델 학습
lr.fit(x_train, y_train)
# → 학습용 문제(X_train)와 정답(y_train)을 이용하여
#   데이터의 관계(회귀식)를 학습


# 3. 학습한 모델로 시험 문제 예측
y_predict = lr.predict(x_test)
# → X_test를 모델에 넣어 연비를 예측
# → 모델이 만든 예상 정답이 y_predict에 저장됨


# 4. 예측 결과 확인
y_predict
````

```
X_train → 공부할 문제

y_test (실제 정답)
    ↕ 비교
y_predict (AI가 예측한 답)

````python
# 5. 모델 평가
score = lr.score(x_test, y_test) # 실제로 점수를 계산해서 모델을 평가하는 코드
print("R² 점수 :", score)
````

| 단계       | 코드                               | 역할                      | 학생 비유         |
| -------- | -------------------------------- | ----------------------- | ------------- |
| 1. 모델 생성 | lr = LinearRegression()        | 선형회귀 모델 준비              | 공부할 학생 준비     |
| 2. 모델 학습 | lr.fit(x_train, y_train)       | 입력값과 결과값의 관계 학습         | 문제와 정답을 보며 공부 |
| 3. 예측    | y_predict = lr.predict(x_test) | 시험 데이터의 연비 예측           | 시험 문제 풀기      |
| 4. 예측값   | y_predict                      | 모델이 예측한 연비              | AI가 작성한 답     |
| 4. 실제값   | y_test                         | 실제 자동차의 연비              | 실제 정답         |
| 5. 모델 평가 | lr.score(x_test, y_test)       | 예측값과 실제값을 비교하여 R² 계산    | 시험 채점         |
| 결과       | R² 점수                          | 모델이 데이터를 얼마나 잘 설명하는지 확인 | 시험 성적         |


```

X_train (공부 문제) + y_train (공부 정답)
              ↓
            fit()
              ↓
         모델이 관계 학습
              ↓
X_test (시험 문제) → predict()
              ↓
       y_predict (AI의 답)
              ↕ 비교
       y_test (실제 정답)
              ↓
         R² 점수(평가)
```
fit()으로 공부 → predict()로 시험 → score()로 채점
